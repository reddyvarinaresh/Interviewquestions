```python
import numpy as np
from scipy.stats import linregress

# Define the data
x = np.array([5, 6, 9, 8, 4])
y = np.array([8, 4, 4, 9, 6])

# Calculate the correlation coefficient
r = np.corrcoef(x, y)[0, 1]

# Calculate the linear regression
slope, intercept, r_value, p_value, std_err = linregress(x, y)

# Print the results
print("Correlation Coefficient:", r)
print("Linear Regression Equation: y =", slope, "* x +", intercept)
```

```python
def find_first_occurrence(arr, element):
    """
    Finds the index of the first occurrence of an element in a list.

    Args:
        arr (list): The input list.
        element: The element to search for.

    Returns:
        int: The index of the first occurrence of the element in the list.
            Returns -1 if the element is not found.
    """
    try:
        index = arr.index(element)
    except ValueError:
        index = -1
    return index


# Example usage
arr = [1, 3, 5, 7, 9, 3, 5, 7, 9]
element = 5
index = find_first_occurrence(arr, element)
if index != -1:
    print(f"The first occurrence of {element} is at index {index}.")
else:
    print(f"{element} is not found in the list.")
```

```python
def list_to_string(lst):
    """
    Converts a list into a string using slicing operator.

    Args:
        lst (list): The input list.

    Returns:
        str: The string representation of the list.
    """
    # Join the list elements with an empty string as separator
    string = ''.join(lst)

    return string


# Example usage
lst = ['H', 'e', 'l', 'l', 'o']
string = list_to_string(lst)
print(f"The list {lst} converted to a string: {string}")
```
